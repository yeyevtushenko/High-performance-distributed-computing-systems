# Завдання 1f. Найпопулярніше жіноче ім'я

> Яке найпопулярніше жіноче ім'я на кораблі? Виділіть з повного імені пасажира (колонка Name) його особисте ім'я (First Name). Спробуйте вручну розібрати кілька значень стовпця Name і виробити правило для вилучення імен, а також поділу їх на жіночі і чоловічі.

```python
print(data['Name'].head(10).to_string())
```

Імена мають вигляд `Прізвище, Титул. Ім'я по батькові`. Для заміжніх жінок часто вказано ім'я чоловіка, а власне ім'я стоїть у дужках, наприклад `Cumings, Mrs. John Bradley (Florence Briggs Thayer)`.

```python
def first_name(name, sex):
    rest = name.split(', ', 1)[1]
    title, after = rest.split('. ', 1)
    if sex == 'female' and title == 'Mrs':
        if '(' not in after:
            return None
        return after.split('(')[1].replace(')', '').split()[0]
    return after.split()[0]

data['FirstName'] = [first_name(n, s) for n, s in zip(data['Name'], data['Sex'])]

female_names = data.loc[data['Sex'] == 'female', 'FirstName'].dropna().value_counts()
male_names = data.loc[data['Sex'] == 'male', 'FirstName'].dropna().value_counts()

print("Жіночі імена:")
print(female_names.head())
print()
print("Чоловічі імена:")
print(male_names.head())

ans_1f = female_names.index[0]
print()
print("Найпопулярніше жіноче ім'я:", ans_1f)
```

Правило таке: для титулів Miss, Ms та інших береться перше слово після титулу. Для Mrs береться перше слово в дужках, бо поза дужками стоїть ім'я чоловіка, тому Mrs без дужок до підрахунку жіночих імен не потрапляють. Для чоловіків береться перше слово після титулу. Стать визначається колонкою `Sex`.

# Завдання 2. Аналіз даних світового рейтингу університетів

Файл рейтингу потрібно завантажити з Kaggle за посиланням із завдання. На Kaggle він має назву `cwurData.csv`, у завданні названий `wurData.csv`, тому код працює з будь-яким із цих імен.

Назви колонок приводяться до нижнього регістру, бо в таблиці завдання поле року подано як `Year`, а у файлі воно може називатися `year`.

# Завдання 2a. Дані лише за 2015 рік

> У рейтингу залиште тільки дані за 2015 рік.

```python
univ = univ[univ['year'] == 2015].copy()
print(univ.shape)
print(univ['year'].unique())
```

# Завдання 2b. Максимальний, мінімальний і середній загальний бал

> Обчисліть максимальний, мінімальний і середній загальний бал світового рейтингу.

```python
score_max = univ['score'].max()
score_min = univ['score'].min()
score_mean = univ['score'].mean()
print("Максимальний бал:", score_max)
print("Мінімальний бал: ", score_min)
print("Середній бал:    ", round(score_mean, 2))
```

# Завдання 2c. Кількість університетів за країнами

> Для всіх згаданих країн виведіть кількість університетів, які увійшли до світового рейтингу. Висновок організуйте за спаданням кількості університетів.

```python
country_counts = univ['country'].value_counts()
print(country_counts.to_string())
```

Метод `value_counts()` одразу повертає результат у порядку спадання.

# Завдання 2d. Сума загальних балів за країнами

> Обчисліть для кожної країни величину що є сумою загальних балів його університетів. Висновок організуйте по спадаючій цієї величини.

```python
country_score_sum = univ.groupby('country')['score'].sum().sort_values(ascending=False)
print(country_score_sum.to_string())
```

# Завдання 2e. Кращий університет для кожної країни

> Обчисліть кращий університет для кожної країни, висновок організуйте по цій величині.

```python
best_by_country = (univ.sort_values('score', ascending=False)
                   .drop_duplicates('country')[['country', 'institution', 'score']]
                   .reset_index(drop=True))
print(best_by_country.to_string())
```

Університети сортуються за загальним балом, і для кожної країни лишається перший, тобто найкращий. Результат упорядкований за балом кращого університету.

# Завдання 2f. Кореляція між двома останніми величинами

> Зробіть висновок про те, чи є кореляція у двох останніх величин.

```python
compare = pd.concat([country_score_sum.rename('sum_score'),
                     best_by_country.set_index('country')['score'].rename('best_score')],
                    axis=1).dropna()

corr_value = compare['sum_score'].corr(compare['best_score'])
print(compare.to_string())
print()
print("Кореляція Пірсона:", round(corr_value, 4))

strength = 'сильна' if abs(corr_value) >= 0.7 else 'помірна' if abs(corr_value) >= 0.3 else 'слабка'
direction = 'прямий' if corr_value > 0 else 'обернений'
ans_2f = f"кореляція {strength}, {direction} зв'язок (r = {corr_value:.2f})"
print("Висновок:", ans_2f)
```

Порівнюються сума балів університетів країни та бал її найкращого університету. Висновок формується за значенням коефіцієнта: від 0.7 за модулем — сильна кореляція, від 0.3 — помірна, нижче — слабка.

# Завдання 2g. П'ять університетів з найбільшим рейтингом

> Обчисліть п'ять університетів з найбільшим рейтингом, виведіть їх назви, загальний бал, країни, рейтинги в своїй країні.

```python
top5 = univ.nlargest(5, 'score')[['institution', 'score', 'country', 'national_rank']]
print(top5.to_string(index=False))
```

# Завдання 2h. П'ять університетів з найкращим рейтингом працевлаштування

> Випускники яких 5 університетів володіють найбільшим робочим рейтингом?

```python
top_alumni = univ.nsmallest(5, 'alumni_employment')[['institution', 'country', 'alumni_employment']]
print(top_alumni.to_string(index=False))
```

Поле `alumni_employment` є рангом працевлаштування випускників, тому кращим вважається найменше значення, тобто найвищий ранг.

# Завдання 2i. Вплив окремих рейтингів на загальний

> Як оцінити вплив окремих рейтингів на загальний. Який із зазначених рейтингів дає найбільший внесок у загальний рейтинг?

```python
rating_columns = ['quality_of_education', 'alumni_employment', 'quality_of_faculty',
                  'publications', 'influence', 'citations', 'broad_impact', 'patents']

ratings = univ[rating_columns].apply(pd.to_numeric, errors='coerce')
correlations = ratings.corrwith(univ['score'])
correlations = correlations.reindex(correlations.abs().sort_values(ascending=False).index)

print(correlations.round(4).to_string())

best_rating = correlations.index[0]
ans_2i = f"{best_rating} (r = {correlations.iloc[0]:.2f})"
print()
print("Найбільший внесок у загальний бал:", ans_2i)
```

Вплив оцінюється кореляцією кожного окремого рейтингу із загальним балом `score`. Оскільки ці показники є рангами, де менше значення краще, кореляція зазвичай від'ємна, тому порівнюються значення за модулем.

**1. Для чого використовується бібліотека Pandas?**

Pandas це пакет Python, який дає швидку, гнучку та виразну структуру даних для того, щоб робота з «реляційними» або «розміченими» даними була легкою та інтуїтивно зрозумілою. **2. Що являє собою Pandas DataFrame?**

DataFrame це один з двох основних об'єктів Pandas, двовимірна таблиця з мітками рядків (індекс) і стовпців. Його можна створити, наприклад, зі словника, де ключі стають іменами стовпців, а значення, задані списками, стають даними. Кожен стовпець DataFrame є об'єктом Series, тобто одновимірним масивом із індексом, і доступ до нього здійснюється за назвою через `df[col]`. Для вибору за позицією використовується `iloc`, а за міткою `loc`.
