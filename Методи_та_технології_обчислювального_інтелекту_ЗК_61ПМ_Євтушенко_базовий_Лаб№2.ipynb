>

# Завдання 14. Векторизація функції maxx

> Перетворіть функцію maxx яка працює з двома скалярними значеннями так, щоб вона могла працювати з масивами (vectorize(maxx, otypes=[float]))
>
> Input:
>
> ```
> def maxx(x, y):
>     """Get the maximum of two items"""
>     if x >= y:
>         return x
>     else:
>         return y
> maxx(1, 5)  #> 5
> ```
>
> Output:
>
> `a = np.array([5, 7, 9, 8, 6, 4, 5])`
>
> `b = np.array([6, 3, 4, 8, 9, 7, 1])`
>
> `pair_max(a, b)`
>
> `#> array([ 6.,  7.,  9.,  8.,  9.,  7.,  5.])`

````python
def maxx(x, y):
    """Get the maximum of two items"""
    if x >= y:
        return x
    else:
        return y

print(maxx(1, 5))

pair_max = np.vectorize(maxx, otypes=[float])

a = np.array([5, 7, 9, 8, 6, 4, 5])
b = np.array([6, 3, 4, 8, 9, 7, 1])
print(repr(pair_max(a, b)))
````

# Завдання 24. Нормалізація sepallength до діапазону [0, 1]

>
> Input:
>
> `url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'`
>
> `sepallength = np.genfromtxt(open_url(url), delimiter=',', dtype='float', usecols=[0])`

````python
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
sepallength = np.genfromtxt(open_url(url), delimiter=',', dtype='float', usecols=[0])

smin = sepallength.min()
smax = sepallength.max()
S = (sepallength - smin) / np.ptp(sepallength)

print(S[:10])
print("min:", S.min(), "max:", S.max())
````

# Завдання 27. Рядки без nan значень

> Виберіть рядки iris_2d датасету які не мають nan значень ([~np.any(np.isnan(row)) for row in iris_2d]).
>
> ```
> url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
> iris_2d = np.genfromtxt(open_url(url), delimiter=',', dtype='float', usecols=[0,1,2,3])
> ```

````python
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
iris_2d = np.genfromtxt(open_url(url), delimiter=',', dtype='float', usecols=[0, 1, 2, 3])
print("Рядків без nan у завантаженому масиві:", len(iris_2d))

np.random.seed(100)
iris_2d[np.random.randint(150, size=20), np.random.randint(4, size=20)] = np.nan

mask = np.array([~np.any(np.isnan(row)) for row in iris_2d])
clean = iris_2d[mask]

print("Рядків після додавання пропусків:", len(iris_2d))
print("Рядків без nan:", len(clean))
print(clean[:5])
````

> Знайдіть кореляцію між SepalLength (перша колонка) та PetalLength (третя колонка) в iris_2d
>
> ```
> url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
> iris_2d = np.genfromtxt(open_url(url), delimiter=',', dtype='float', usecols=[0,1,2,3])
> ```

````python
url = 'https://archive.ics.uci.edu/ml/machine-learning-databases/iris/iris.data'
iris_2d = np.genfromtxt(open_url(url), delimiter=',', dtype='float', usecols=[0, 1, 2, 3])

corr = np.corrcoef(iris_2d[:, 0], iris_2d[:, 2])[0, 1]
print("Кореляція Пірсона:", corr)
````

Значення близьке до 1 означає сильний прямий зв'язок: чим довша чашолистка, тим довша й пелюстка.

# Завдання 30. Піки в 1D масиві

> Знайдіть всі піки в 1D numpy масиві a. Піки це точки з двох боків які мають менших сусідів (diff, where, sign).
>
> Input: `a = np.array([1, 3, 7, 1, 2, 6, 0, 1])`
>
> Output: `#> array([2, 5])`
>
> Де 2 та 5 це позиції піків 7 та 6

````python
a = np.array([1, 3, 7, 1, 2, 6, 0, 1])

double_diff = np.diff(np.sign(np.diff(a)))
peak_locations = np.where(double_diff == -2)[0] + 1

print(repr(peak_locations))
print("Значення в піках:", a[peak_locations])
````

Перший `np.diff` показує, зростає чи спадає послідовність, `np.sign` залишає лише знак, а другий `np.diff` дає `-2` там, де зростання змінюється спаданням. До позицій додається 1, бо друга різниця коротша за вихідний масив на два елементи.

Основні математичні функції працюють поелементно одразу на всьому масиві, а broadcasting дозволяє виконувати операції над масивами різної форми без явних циклів, що робить код стислішим і швидшим. **Недоліки numpy.** Усі елементи масиву мають бути одного типу. Зріз масиву є поданням тих самих даних, тому зміна зрізу змінює й оригінал. Оператор `*` виконує поелементне множення, а не матричне, і для матричного потрібна функція `dot`. Транспонування одновимірного масиву нічого не змінює, а змішування цілочисельної індексації зі зрізами знижує ранг результату. Крім того, для багатовимірних масивів треба вказувати діапазон для кожного виміру.

# Контрольні питання

**1. Які основні операції із матрицями реалізовано у бібліотеці numpy?**

Основні математичні операції (додавання, віднімання, множення, ділення та квадратний корінь) виконуються поелементно, і доступні як оператори `+`, `-`, `*`, `/`, так і як функції `np.add`, `np.subtract`, `np.multiply`, `np.divide`, `np.sqrt`. Для скалярного добутку векторів, множення вектора на матрицю та множення матриць використовується функція `dot`, яка є і функцією модуля, і методом масиву. Також є `np.sum` для суми всіх елементів або суми за рядками чи стовпцями через `axis`, транспонування через атрибут `T` та broadcasting, який дозволяє працювати з масивами різної форми.

**2. Яка бібліотека використовується для побудови рисунків у Python?**

Для побудови графіків використовується бібліотека Matplotlib. Її найважливіша функція `plot` дозволяє зображувати 2D дані, а `show` виводить рисунок на екран. Також можна будувати кілька графіків одразу, додавати назву, легенду та підписи осей через `title`, `legend`, `xlabel` і `ylabel`, розміщувати кілька графіків на одному рисунку через `subplot` і показувати зображення функцією `imshow`.
